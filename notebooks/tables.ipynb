```python
from pathlib import Path
import pandas as pd
from IPython.display import display

# Works whether the notebook kernel starts in the project root or notebooks/.
candidates = [Path.cwd() / "data-source" / "data", Path.cwd().parent / "data-source" / "data"]
DATA_DIR = next((path for path in candidates if path.exists()), None)
if DATA_DIR is None:
    raise FileNotFoundError("Could not find data-source/data")

seasons = sorted(
    path.name for path in DATA_DIR.iterdir()
    if path.is_dir() and path.name[:4].isdigit()
)
print("Available seasons:", seasons)
```

```python
SEASON = seasons[-1]  # Example: "2025-26"
season_dir = DATA_DIR / SEASON
print("Selected season:", SEASON)
```

## Main tables for the selected season

```python
tables = {
    "players_raw": season_dir / "players_raw.csv",
    "cleaned_players": season_dir / "cleaned_players.csv",
    "fixtures": season_dir / "fixtures.csv",
    "teams": season_dir / "teams.csv",
    "player_idlist": season_dir / "player_idlist.csv",
    "merged_gw": season_dir / "gws" / "merged_gw.csv",
    "gw1": season_dir / "gws" / "gw1.csv",
    "xP1": season_dir / "gws" / "xP1.csv",
}

loaded_tables = {}
for name, path in tables.items():
    print(f"\n--- {name} ---")
    if path.exists():
        loaded_tables[name] = pd.read_csv(path)
        df = loaded_tables[name]
        print(f"{len(df):,} rows x {len(df.columns)} columns")
        display(df.head())
    else:
        print(f"Missing: {path}")
```

## Global merged tables

```python
global_tables = {
    "cleaned_merged_seasons": DATA_DIR / "cleaned_merged_seasons.csv",
    "cleaned_merged_seasons_team_aggregated": DATA_DIR / "cleaned_merged_seasons_team_aggregated.csv",
    "master_team_list": DATA_DIR / "master_team_list.csv",
}

for name, path in global_tables.items():
    print(f"\n--- {name} ---")
    if path.exists():
        df = pd.read_csv(path)
        print(f"{len(df):,} rows x {len(df.columns)} columns")
        display(df.head())
    else:
        print(f"Missing: {path}")
```
